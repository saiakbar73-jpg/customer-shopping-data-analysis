```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create customer shopping data
data = {
    "Customer": ["Ayesha", "Rahul", "Priya", "Aman", "Sara"],
    "Product": ["Laptop", "Mobile", "Headphones", "Laptop", "Mobile"],
    "Category": ["Electronics", "Electronics", "Accessories", "Electronics", "Electronics"],
    "City": ["Nagpur", "Pune", "Nagpur", "Mumbai", "Pune"],
    "Sales": [55000, 25000, 2300, 55000, 25000]
}

# Step 2: Create a DataFrame
df = pd.DataFrame(data)

# Step 3: Display the dataset
print("\nCUSTOMER SHOPPING DATA")
print(df)

# Step 4: Calculate total sales
print("\nTotal Sales:", df["Sales"].sum())

# Step 5: Calculate average sales
print("Average Sales:", df["Sales"].mean())

# Step 6: Find the highest sale
print("Highest Sale:", df["Sales"].max())

# Step 7: Find the lowest sale
print("Lowest Sale:", df["Sales"].min())
```

```python
# Step 9: Product-wise sales analysis

product_sales = df.groupby("Product")["Sales"].sum()

print("\nProduct-wise Sales:")
print(product_sales)

# Find the best-selling product
best_product = product_sales.idxmax()

print("\nBest-Selling Product:", best_product)
```

```python
# Step 10: Visualize product-wise sales

import matplotlib.pyplot as plt

product_sales.plot(kind="bar")

plt.title("Product-wise Sales Analysis")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Step 11: Count customers in each city

customer_count = df.groupby("City")["Customer"].count()

print("\nNumber of Customers in Each City:")
print(customer_count)
```

```python
customers_above_25000 = df[df["Sales"] > 25000]

print(customers_above_25000)
```

```python
nagpur_customers = df[df["City"] == "Nagpur"]

print(nagpur_customers)
```

```python
customer_count = df["City"].value_counts()

print(customer_count)
```

```python
customer_count.plot(kind="bar", title="Number of Customers by City")

plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.show()
```

```python
top_3_customers = df.nlargest(3, "Sales")

print(top_3_customers)
```

```python
total_sales = df["Sales"].sum()

df["Sales_Percentage"] = (df["Sales"] / total_sales) * 100

print(df[["Customer", "Sales", "Sales_Percentage"]])
```
